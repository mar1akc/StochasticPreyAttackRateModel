```python
"""Exact L1,L2,L3 and det M for the expOU driver"""
import sympy as sp
X,Y1,Y2,U=sp.symbols('X Y1 Y2 U',real=True)
r,a,k,b,D,v1,v2,s=sp.symbols('r a k b D v1 v2 s',positive=True)
al,muU,sU,gm,Dl=sp.symbols('alpha mu_U sigma_U g_minus Delta',positive=True)
z=[X,Y1,Y2,U]; G=gm+Dl*sp.exp(U)
bv=[X*(r-a*X+s*Y1-b*Y2), k*X*Y2-Y1*(G*X+D+v1), D*Y1-v2*Y2, al*(muU-U)]
sig=[0,0,0,sU]
def lie(u,w):
    return [sp.expand(sum(u[j]*sp.diff(w[i],z[j])-w[j]*sp.diff(u[i],z[j]) for j in range(4))) for i in range(4)]
L1=lie(bv,sig); L2=lie(bv,L1); L3=lie(bv,L2)
print("L1:")
for i,n in enumerate(['X','Y1','Y2','U']): print(f"  {n}: {sp.simplify(L1[i])}")
print("\nL2:")
for i,n in enumerate(['X','Y1','Y2','U']): print(f"  {n}: {sp.factor(sp.simplify(L2[i]))}")
print("\nL3 X-comp factored:"); print(" ",sp.factor(sp.simplify(L3[0])))
print("\nL3 Y1-comp factored:"); print(" ",sp.factor(sp.simplify(L3[1])))
print("\nL3 Y2-comp factored:"); print(" ",sp.factor(sp.simplify(L3[2])))
print("\nL3 U-comp:", sp.simplify(L3[3]))
# Q_X and Q_Y2 in the form L3_X = sU*Dl*e^U*X^2*Q_X ; L3_Y2 = -sU*Dl*e^U*D*X*Q_Y2
QX=sp.simplify(L3[0]/(sU*Dl*sp.exp(U)*X**2))
QY1=sp.simplify(L3[1]/(sU*Dl*sp.exp(U)*X))
QY2=sp.simplify(-L3[2]/(sU*Dl*sp.exp(U)*D*X))
print("\nQ_X =",sp.expand(QX))
print("\nQ_Y1 =",sp.expand(QY1))
print("\nQ_Y2 =",sp.expand(QY2))
print("\ns*Q_Y2 + Q_X =",sp.simplify(sp.expand(s*QY2+QX)))
M=sp.Matrix.hstack(*[sp.Matrix(v) for v in [sig,L1,L2,L3]])
print("\ndet M =",sp.factor(sp.expand(M.det())))
```
